```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
import time

def open_99acres():
    """
    Initializes a Chrome driver and navigates to 99acres.com.
    """
    # 1. Setup Chrome Options to look less like a bot
    chrome_options = Options()
    chrome_options.add_argument("--start-maximized")
    chrome_options.add_argument("--disable-blink-features=AutomationControlled")
    # Optional: chrome_options.add_argument("--headless") # Run without a popup window
    
    # 2. Automatically manage the ChromeDriver version
    service = Service(ChromeDriverManager().install())
    
    # 3. Initialize the driver
    driver = webdriver.Chrome(service=service, options=chrome_options)
    
    try:
        print("Navigating to 99acres...")
        driver.get("https://www.99acres.com/")
        
        # Give the page a few seconds to load its JavaScript components
        time.sleep(5) 
        
        print(f"Successfully loaded: {driver.title}")
        return driver

    except Exception as e:
        print(f"An error occurred: {e}")
        driver.quit()
        return None

def perform_city_search(driver, city_name):
    """
    Clicks the search bar, types the city, and selects the first suggestion.
    """
    wait = WebDriverWait(driver, 15)
    
    try:
        # 1. Wait for the search input to be present and click it
        # The 'keyword' ID is the standard search wrapper on 99acres
        search_box = wait.until(EC.element_to_be_clickable((By.ID, "keyword")))
        search_box.click()
        print(f"Clicked search bar. Typing '{city_name}'...")

        # 2. Type the city name
        search_box.send_keys(city_name)

        # 3. Wait for the suggestion dropdown to appear and click the first result
        # The suggestions usually appear in a div with a specific class or list items
        suggestion_xpath = "//div[contains(@class, 'suggest')]//li" 
        first_suggestion = wait.until(EC.element_to_be_clickable((By.XPATH, suggestion_xpath)))
        
        print("Selecting from dropdown...")
        first_suggestion.click()

        # 4. Optional: If clicking the suggestion doesn't auto-redirect, hit the Search button
        # Usually, 99acres has a 'Search' button with a specific class
        # search_button = driver.find_element(By.CLASS_NAME, "search_btn")
        # search_button.click()

        print(f"Search for {city_name} executed successfully.")
        
    except Exception as e:
        print(f"Error during search: {e}")

# Execute the function
if __name__ == "__main__":
    browser = open_99acres()
    
    # Keep it open for a bit so you can see it worked
    if browser:
        time.sleep(10)
        browser.quit()
```
